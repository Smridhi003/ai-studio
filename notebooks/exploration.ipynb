```python
import sys
sys.path.insert(0, '..')

from utils.nlp_analyzer import NLPAnalyzer
from utils.sentiment import SentimentAnalyzer
from utils.classifier import ZeroShotClassifier
from utils.generator import TextGenerator
from utils.neural_net import NeuralNetSimulator

print('✅ All modules loaded')
```

## 1. NLP Analysis — Named Entity Recognition

```python
nlp = NLPAnalyzer()
text = 'Elon Musk founded SpaceX in 2002 in Hawthorne, California. Tesla reported $25 billion in revenue.'
result = nlp.analyze(text)

print('Entities:')
for e in result['entities']:
    print(f"  [{e['type']}] {e['text']}")

print(f"\nTokens: {result['tokens']}")
print(f"Sentences: {result['sentences']}")
print(f"Key phrases: {result['key_phrases']}")
print(f"Readability: {result['readability']}")
```

## 2. Sentiment Analysis — Emotion Detection

```python
sa = SentimentAnalyzer()
texts = [
    'This is absolutely amazing and I love it!',
    'Terrible experience. Worst purchase ever.',
    'It was okay, nothing special.',
    'Brilliant product but terrible delivery.',
]
for t in texts:
    r = sa.analyze(t)
    print(f"Text: '{t[:50]}...'")
    print(f"  Overall: {r['overall']} (score: {r['score']:.2f})")
    top_emotion = max(r['emotions'], key=r['emotions'].get)
    print(f"  Top emotion: {top_emotion} ({r['emotions'][top_emotion]:.2f})\n")
```

## 3. Zero-Shot Classification

```python
clf = ZeroShotClassifier()
text = 'Scientists at MIT have developed a new battery technology for electric vehicles.'
labels = ['Technology', 'Science', 'Business', 'Sports', 'Health']

result = clf.classify(text, labels)
print(f"Text: {text}\n")
print('Scores:')
for label, score in sorted(result['scores'].items(), key=lambda x: -x[1]):
    bar = '█' * int(score * 20)
    print(f"  {label:<15} {bar} {score:.2%}")
print(f"\nTop: {result['top']}")
```

## 4. Text Generation

```python
gen = TextGenerator()
topic = 'artificial intelligence in healthcare'

for style in ['news', 'eli5', 'summary']:
    result = gen.generate(topic, style=style, length='short')
    print(f"=== Style: {style.upper()} ===")
    print(result['text'][:300])
    print(f"(Model: {result['model']})\n")
```

## 5. Neural Network — Forward Pass Visualization

```python
nn = NeuralNetSimulator()

sentences = [
    'This is absolutely fantastic!',
    'I hate this terrible product.',
    'The weather is okay today.',
]

for sent in sentences:
    r = nn.forward_pass(sent)
    print(f"Input: '{sent}'")
    print(f"  Features: {r['input_features']}")
    print(f"  Prediction: {r['prediction']} ({r['confidence']:.1%} confidence)")
    print(f"  Probs: {r['probabilities']}\n")
```
